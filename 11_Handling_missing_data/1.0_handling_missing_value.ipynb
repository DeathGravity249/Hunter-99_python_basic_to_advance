```python
## Imputation missing values
## 1->Mean values Imputation
from matplotlib import pyplot as plt

sn.histplot(df["age"],kde=True)
plt.show()
```

```python
df["age_mean"]=df["age"].fillna(df["age"].mean())
```

```python
df[["age_mean","age"]]

## It returns a new DataFrame containing only those two columns.

## If either "age_mean" or "age" does not exist, it will raise a KeyError.
```

```python
## Mean imputation works well when we have normally distributed data 
```

```python
## Meadin value Imputation if we have outlier in the dataset 
df["age_meadian"]=df["age"].fillna(df["age"].median())
```

```python
df[["age_meadian","age"]]
```

```python
## Mode imputation techniques -- categorical values
df[df["embarked"].isnull()]
```

df[ ... ]:
Filters the original DataFrame using the boolean mask. So it only keeps rows where the condition is True.

```python
df[df["embarked"].isnull()]
```

```python
mode_value=df[df["embarked"].notna()]["embarked"].mode()[0]
```

```python
## This is crating a new column 

df["embark_mode"]=df["embarked"].fillna(mode_value)
```

```python
df[["embark_mode","embarked"]]
```
